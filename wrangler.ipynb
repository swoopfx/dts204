```python
#Retrieve the unique counts of string availaible 
#We use it for Standardization and mapping 
df["Country"].value_counts()
```

```python
# Remove leading/trailing whitespace
df["Country"] = df["Country"].str.strip()

# Standardize the Country column
country_mapping = {
    "United States": "United States",
    "U.S.A.": "United States",
    "usa": "United States"
}

# Replace values
df["Country"] = df["Country"].replace(country_mapping)
```

```python
# Ensure Total Amount is numeric
df["Total Amount"] = pd.to_numeric(df["Total Amount"], errors="coerce")
```

```python
#Confirm that the Total Amount Column does not have NaN
df.info()
```

```python
# Remove rows with missing revenue values if it exist from previous action
df = df.dropna(subset=["Total Amount"])
```

```python
# Aggregate total revenue by product category
df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)
```

```python





revenue_by_category = (
   df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)
)

import matplotlib.pyplot as plt
# Plot the bar chart
plt.figure(figsize=(8, 5))
revenue_by_category.plot(kind="bar")

plt.title("Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
